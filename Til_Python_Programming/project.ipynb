## Objective

The objective of this project is to assess the extent to which the public charging infrastructure in Dutch municipalities corresponds with the growth of electric passenger cars. Municipal-level electric vehicle (EV) registration data from 2022 to 2025 is combined with charging-point availability in 2025 to evaluate whether municipalities where the EV fleets grew faster or are larger also exhibit higher infrastructure provision in the form of public charging stations. The analysis aims to identify relationships between EV private use and public charging capacity, and to determine whether the public charging infrastructure aligns with the potential demand based on the number of electric passenger cars.

## Research question

**To what extent does the public charging infrastructure in Dutch municipalities in 2025 correspond to the number and growth of electric passenger cars between 2022 and 2025?**

Sub-questions:
1. How has the number of electric passenger cars developed per municipality between 2022 and 2025?
2. How are public charging points distributed across Dutch municipalities in 2025?
3. How does the number of electric passenger cars per public charging point vary across municipalities, and which municipalities have the highest and lowest ratios?

## Geographical and temporal scale

The study is conducted at the municipal level within the Netherlands, examining differences in public charging infrastructure and electric vehicle adoption across Dutch municipalities.

The analysis covers the period 2022 to 2025, enabling assessment of short-term developments in both charging infrastructure deployment and the increase in electric passenger cars.

# Data Pipeline

1. **Retrieve geographical coordinates for all public charging stations.** These coordinates form the basis for spatially assigning each charging point to the correct municipality. Check whether the data counts charging stations as a whole or individual charging points.
2. **Obtain and define municipal boundary data.** Municipal boundaries are used to spatially link charging-station locations to municipalities.
3. **Determine the number of charging stations per municipality** by intersecting charging-station coordinates with municipal boundaries.
4. **Clean and prepare the dataset containing electric-vehicle registrations per municipality.** This includes harmonising municipality names and ensuring consistent temporal coverage.
5. **Verify municipality alignment across datasets.** Municipalities that appear in the EV dataset but have no public charging stations are assigned a charging-station count of zero.
6. **Calculate EV growth per municipality from 2022 to 2025**, expressed in absolute numbers and percentages.
7. **Compare EV growth and current EV levels with the number of charging stations in each municipality.** This shows whether municipalities with higher EV adoption or faster growth also have more charging infrastructure available in 2025.
