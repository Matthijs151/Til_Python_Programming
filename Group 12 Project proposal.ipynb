# Introduction

# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

The objective of this project is to assess the extent to which the growth of public charging infrastructure in Dutch municipalities corresponds with the growth of electric passenger cars. Municipal-level electric vehicle (EV) registration data from 2022 to 2025 is combined with data on public charging-point availability over the same period. The analysis examines whether municipalities with larger or faster-growing EV fleets also have higher levels of public charging infrastructure and infrastructure growth. In this way, the project aims to identify the relationship between private EV adoption and public charging infrastructure, and to assess whether infrastructure provision is aligned with potential charging demand based on the number of electric passenger cars.

# Research question

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

To what extent does the provision of public charging infrastructure in Dutch municipalities correspond to the number and growth of electric passenger cars between 2022 and 2025?

Sub questions:

1. How has the number of electric passenger cars developed across Dutch municipalities between 2022 and 2025?
2. How has the provision of public charging infrastructure developed across Dutch municipalities between 2022 and 2025?
3. To what extent does the provision of public charging infrastructure correspond to the number and growth of electric passenger cars across Dutch municipalities?


# Data Used

# Data Pipeline

1. Intended data-analyses pipeline 
Retrieve geographical coordinates for all public charging stations.  
These coordinates form the basis for spatially assigning each charging point to the correct municipality. 
analyse if data uses charging stations as a whole or also count charging points
2. Obtain and define municipal boundary data.  
Municipal boundaries are used to spatially link charging‑station locations to municipalities.
3. Determine the number of charging stations per municipality  
by intersecting charging‑station coordinates with municipal boundaries.
4. Clean and prepare the dataset containing electric‑vehicle registrations per municipality. Includes harmonizing municipality names and ensuring consistent temporal coverage.
5. Verify municipality alignment across datasets.  
Municipalities that appear in the EV dataset but have no public charging stations will be assigned a charging‑station count of zero.
6. Calculate EV growth per municipality from 2022 to 2025.  
Growth will be expressed in absolute numbers or percentages
7. Compare EV growth and current EV levels with the number of charging stations in each municipality.  
This allows assessment of whether municipalities with higher EV adoption or faster growth also have more charging infrastructure available in 2025.


# Geographical scale and temporal scale

The study is conducted at the municipal level within the Netherlands, examining differences in public charging infrastructure and electric vehicle adoption across Dutch municipalities. 
The analysis covers the period 2022 to 2025, enabling assessment of short‑term developments in both charging infrastructure deployment and the increase in electric passenger cars.
