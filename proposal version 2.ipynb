# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

The objective of this project is to assess how the spatial distribution of public charging infrastructure in 2025 across Dutch municipalities aligns with potential demand factors. Because charging infrastructure data prior to 2025 is unavailable, the analysis evaluates the 2025 charging network. It examines how this current capacity is explained by past electric vehicle (EV) growth (2022–2025) per municipality and municipal population density. By incorporating population density as a key component, this study aims to identify whether charging infrastructure deployment responds primarily to localized vehicle adoption trends or broader urban density characteristics.

# Research question

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

To what extent is the public charging infrastructure in Dutch municipalities explained by the growth of electric passenger cars (2022-2025) and municipal population density?


Sub questions:

1. How has the total number of electric passenger cars developed within Dutch municipalities between 2022 and 2025? 
2. How does the spatial density of public charging points in 2025 correlate with municipal population density?
3. How does the ratio of electic passenger cars to public charging points in 2025 vary across municipalities and how do population density and past EV growth rates explain this variation?

# Data Pipeline


1. Retrieve Geospatial & EV Data: Retrieve geographical coordinates for all 2025 public charging stations and municipal boundary files. Fetch CBS EV registration datasets (2022–2025).

2. Integrate Population Density Data: Obtain municipal-level population density data.

3. Spatial Aggregation: Map individual charging station coordinates into municipal boundaries to calculate total charging capacity/points per municipality for 2025.

4. Harmonization & Cleaning: Harmonize municipality names and codes across the EV registration, population density, and charging point datasets to ensure consistency.

5. Feature Engineering: Calculate absolute and percentage EV growth rates per municipality from 2022 to 2025. Compute the EV-to-charging-point ratios for 2025.

6. Statistical & Spatial Modeling: Run a multivariate regression or correlation analysis where the 2025 charging point density serves as the dependent variable, and 2025 EV volume, 2022–2025 EV growth, and population density serve as the explanatory independent variables.

# Geographical scale and temporal scale

The study is conducted at the municipal level within the Netherlands, examining differences in public charging infrastructure and electric vehicle adoption across Dutch municipalities. 
The analysis covers the period 2022 to 2025, enabling assessment of short‑term developments in both charging infrastructure deployment and the increase in electric passenger cars.
